```python
from pathlib import Path
import zipfile
import shutil
from pyprojroot import here
```

```python
zip_path = here() / "data" / "clouds-photos.zip"
extract_dir = here() / "data"
```

```python
with zipfile.ZipFile(zip_path, "r") as z:
    z.extractall(extract_dir)
```

```python
dst = extract_dir / "clouds"
dst.mkdir(exist_ok=True)
```

```python
for split in ["clouds_train", "clouds_test"]:
    split_dir = extract_dir / split
    for class_dir in split_dir.iterdir():
        if class_dir.is_dir():
            for img in class_dir.glob("*.jpg"):
                shutil.copy2(img, dst / img.name)
```

```python
zip_path.unlink()                            # delete clouds-photos.zip
shutil.rmtree(extract_dir / "clouds_train")
shutil.rmtree(extract_dir / "clouds_test")
```

```python
from PIL import Image

import matplotlib.pyplot as plt

img_path = list(dst.glob("*.jpg"))[0]

img = Image.open(img_path)
plt.figure(figsize=(8, 6))
plt.imshow(img)
plt.axis('off')
plt.title(img_path.name)
plt.tight_layout()
plt.show()
```
