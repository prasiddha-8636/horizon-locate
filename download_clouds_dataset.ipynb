```python
from pathlib import Path
import zipfile
import shutil
```

```python
zip_path = Path("data/clouds-photos.zip")
extract_dir = Path("data")
```

```python
with zipfile.ZipFile(zip_path, "r") as z:
    z.extractall(extract_dir)
```

```python
dst = extract_dir / "clouds"
dst.mkdir(exist_ok=True)
```

```python
for split in ["clouds_train", "clouds_test"]:
    split_dir = extract_dir / split
    for class_dir in split_dir.iterdir():
        if class_dir.is_dir():
            for img in class_dir.glob("*.jpg"):
                shutil.copy2(img, dst / img.name)
```

```python
zip_path.unlink()                     # delete clouds-photos.zip
shutil.rmtree(extract_dir / "clouds_train")
shutil.rmtree(extract_dir / "clouds_test")
```
