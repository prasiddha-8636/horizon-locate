```python
import os
import numpy as np
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
import torchvision.transforms as T
import torchvision.transforms.functional as TF
from PIL import Image
import segmentation_models_pytorch as smp
from tqdm import tqdm
```

### 2. Configuration and Device Setup

Here we define the device for training (GPU if available, otherwise CPU) and set up the directory paths for images and masks. We also define the ImageNet mean and standard deviation for normalization.

```python
# Start training flow on GPU
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Running training on device: {device}")

# Configure directories
syn_img_dir = "data/synthetic_dataset/images"
syn_mask_dir = "data/synthetic_dataset/masks"

# ImageNet normalization parameters
imagenet_mean = [0.485, 0.456, 0.406]
imagenet_standard_deviation = [0.229, 0.224, 0.225]
```

### 3. Defining Dataset Class

The `UnifiedDatasetAug` class handles loading images and masks, applying transformations (including data augmentation for training), and preparing them for the model. It includes haze simulation and scaling variations, crucial for high-altitude domain adaptation.

```python
class UnifiedDatasetAug(Dataset):
    def __init__(self, imgs, masks, is_train=True):
        self.imgs = imgs
        self.masks = masks
        self.is_train = is_train

        if self.is_train:
            self.img_trans = T.Compose([
                T.Resize(256),
                T.CenterCrop(256),
                T.ColorJitter(brightness=0.3, contrast=0.3, saturation=0.3, hue=0.05),
                T.GaussianBlur(kernel_size=5, sigma=(0.5, 2.0)),
                T.ToTensor(),
                T.Normalize(mean=imagenet_mean, std=imagenet_standard_deviation)
            ])
        else:
            self.img_trans = T.Compose([
                T.Resize(256),
                T.CenterCrop(256),
                T.ToTensor(),
                T.Normalize(mean=imagenet_mean, std=imagenet_standard_deviation)
            ])

        self.mask_trans = T.Compose([
            T.Resize(256, interpolation=T.InterpolationMode.NEAREST),
            T.CenterCrop(256),
            T.ToTensor()
        ])

    def __len__(self):
        return len(self.imgs)

    def __getitem__(self, idx):
        img_pil = Image.open(self.imgs[idx]).convert("RGB")
        mask_pil = Image.open(self.masks[idx]).convert("L")
        
        if self.is_train and np.random.rand() > 0.5:
            img_pil = TF.hflip(img_pil)
            mask_pil = TF.hflip(mask_pil)
    
        img = self.img_trans(img_pil)
        mask = self.mask_trans(mask_pil)
        return img, mask
```

### 4. IoU Computation Function

This helper function calculates the Intersection over Union (IoU), a common metric for evaluating segmentation models.

```python
def compute_iou(pred_logits, true_masks, threshold=0.5):
    preds = (torch.sigmoid(pred_logits) > threshold).float()
    intersection = (preds * true_masks).sum()
    union = preds.sum() + true_masks.sum() - intersection
    return (intersection / (union + 1e-6)).item()
```

### 5. Data Splitting and DataLoader Initialization

Here, the dataset is split into training and validation sets, and `DataLoader` instances are created for efficient batch processing during training.

```python
train_images, train_masks = [], []
val_images, val_masks = [], []

if os.path.exists(syn_img_dir):
    all_files = sorted([f for f in os.listdir(syn_img_dir) if f.endswith(".png")])

    # 80/20 train/validation split of your 500 photorealistic images
    train_files = all_files[:400]
    val_files = all_files[400:]

    for f in train_files:
        train_images.append(os.path.join(syn_img_dir, f))
        train_masks.append(os.path.join(syn_mask_dir, f))

    for f in val_files:
        val_images.append(os.path.join(syn_img_dir, f))
        val_masks.append(os.path.join(syn_mask_dir, f))

train_dataset = UnifiedDatasetAug(train_images, train_masks, is_train=True)
val_dataset = UnifiedDatasetAug(val_images, val_masks, is_train=False)

train_loader = DataLoader(train_dataset, batch_size=8, shuffle=True, num_workers=2)
val_loader = DataLoader(val_dataset, batch_size=8, shuffle=False, num_workers=2)

print(f"Number of training images: {len(train_images)}")
print(f"Number of validation images: {len(val_images)}")
```

### 6. Model Definition

We define the U-Net model with a `tu-mobilenetv3_large_100` encoder, pre-trained on ImageNet. The model is moved to the specified device (GPU if available).

```python
model = smp.Unet(
    encoder_name="tu-mobilenetv3_large_100",
    encoder_weights="imagenet",
    in_channels=3,
    classes=1
).to(device)
```

### 7. Loss Function and Optimizer

The `BCEWithLogitsLoss` is used as the loss function, and `AdamW` is chosen as the optimizer.

```python
criterion = nn.BCEWithLogitsLoss()
optimizer = torch.optim.AdamW(model.parameters(), lr=1e-4)
```

### 8. Training Loop

This section contains the main training loop, iterating for a specified number of epochs. It handles forward passes, backpropagation, optimization, and evaluation of both training and validation performance. The model's state dictionary is saved to Google Drive if a new best validation IoU is achieved.

```python
epochs = 15 # Transfer learning only requires 15 epochs to converge on 500 samples
best_val_iou = 0.0

for epoch in range(epochs):
    model.train()
    total_loss = 0.0
    total_iou = 0.0

    for images, masks in tqdm(train_loader, desc=f"Epoch {epoch+1}/{epochs} (Train)"):
        images, masks = images.to(device), masks.to(device)
        optimizer.zero_grad()
        outputs = model(images)
        loss = criterion(outputs, masks)
        loss.backward()
        optimizer.step()

        total_loss += loss.item()
        total_iou += compute_iou(outputs, masks)

    train_loss = total_loss / len(train_loader)
    train_iou = total_iou / len(train_loader)

    model.eval()
    val_loss = 0.0
    val_iou = 0.0

    with torch.no_grad():
        for images, masks in tqdm(val_loader, desc=f"Epoch {epoch+1}/{epochs} (Val)"):
            images, masks = images.to(device), masks.to(device)
            outputs = model(images)
            loss = criterion(outputs, masks)
            val_loss += loss.item()
            val_iou += compute_iou(outputs, masks)

    v_loss = val_loss / len(val_loader)
    v_iou = val_iou / len(val_loader)

    print(f"Epoch {epoch+1}/{epochs} Summary:")
    print(f" -> Train Loss: {train_loss:.4f} | Train IoU: {train_iou*100.0:.2f}%")
    print(f" -> Val Loss:   {v_loss:.4f} | Val IoU:   {v_iou*100.0:.2f}%")

    # Save the best model parameters permanently to Google Drive
    if v_iou > best_val_iou:
        best_val_iou = v_iou
        torch.save(model.state_dict(), "/content/drive/MyDrive/sky_segmentation_unet_model.pth")
        print(f" ==> Checkpoint saved to Google Drive with new best Val IoU: {best_val_iou*100.0:.2f}%")

print(f"\nTraining completed! Checkpoint stored on Drive. Highest Val IoU: {best_val_iou*100.0:.2f}%")
```
